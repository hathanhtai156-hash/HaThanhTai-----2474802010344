## Câu 7: Tạo biến TBM1, TBM2, TBM3

```python
# TBM1: điểm trung bình lớp 10
# TBM2: điểm trung bình lớp 11
# TBM3: điểm trung bình lớp 12

def tinh_tbm_hoc_ky(ky):
    return (
        df[f"T{ky}"] * 2 +
        df[f"L{ky}"] +
        df[f"H{ky}"] +
        df[f"S{ky}"] +
        df[f"V{ky}"] * 2 +
        df[f"X{ky}"] +
        df[f"D{ky}"] +
        df[f"N{ky}"]
    ) / 10

# Tính điểm trung bình từng học kỳ
tbm_hk1 = tinh_tbm_hoc_ky(1)
tbm_hk2 = tinh_tbm_hoc_ky(2)
tbm_hk3 = tinh_tbm_hoc_ky(3)
tbm_hk4 = tinh_tbm_hoc_ky(4)
tbm_hk5 = tinh_tbm_hoc_ky(5)
tbm_hk6 = tinh_tbm_hoc_ky(6)

# Tính điểm trung bình từng năm
# HK2 nhân hệ số 2
df["TBM1"] = ((tbm_hk1 + tbm_hk2 * 2) / 3).round(2)
df["TBM2"] = ((tbm_hk3 + tbm_hk4 * 2) / 3).round(2)
df["TBM3"] = ((tbm_hk5 + tbm_hk6 * 2) / 3).round(2)

print("KẾT QUẢ TẠO BIẾN TBM1, TBM2, TBM3")
display(df[["TBM1", "TBM2", "TBM3"]].head(10))

print("Thống kê mô tả của TBM1, TBM2, TBM3:")
display(df[["TBM1", "TBM2", "TBM3"]].describe())

print("Kết luận: Đã tạo thành công 3 biến TBM1, TBM2, TBM3 cho lớp 10, 11, 12.")
```

## 8. Tạo các biến xếp loại XL1, XL2 và XL3 dựa trên TBM1,TBM2 và TBM3 cho từng năm lớp 10, 11, 12

```python
# Hàm xếp loại học lực
def xep_loai(tbm):
    if tbm < 5.0:
        return "Y"      # Yếu
    elif tbm < 6.5:
        return "TB"     # Trung bình
    elif tbm < 8.0:
        return "K"      # Khá
    elif tbm < 9.0:
        return "G"      # Giỏi
    else:
        return "XS"     # Xuất sắc

# Tạo các biến xếp loại
df["XL1"] = df["TBM1"].apply(xep_loai)
df["XL2"] = df["TBM2"].apply(xep_loai)
df["XL3"] = df["TBM3"].apply(xep_loai)

# Hiển thị kết quả 10 dòng đầu
print("KẾT QUẢ TẠO BIẾN XL1, XL2, XL3")
display(df[["TBM1", "XL1", "TBM2", "XL2", "TBM3", "XL3"]].head(10))

# Thống kê tần số xếp loại
print("Tần số xếp loại XL1:")
display(df["XL1"].value_counts())

print("Tần số xếp loại XL2:")
display(df["XL2"].value_counts())

print("Tần số xếp loại XL3:")
display(df["XL3"].value_counts())

print("Kết luận: Đã tạo thành công các biến xếp loại XL1, XL2, XL3.")
```

## 9. Tạo các biến US_TBM1, US_TBM2 và US_TBM3 để chuyển điểm trung bình các năm lớp 10, 11 và 12 từ thang điểm 10 của Việt Nam sang thang điểm 4 của Mỹ. Sử dụng phương pháp Min-Max Normalization 

```python
# Chuyển điểm trung bình từ thang điểm 10 sang thang điểm 4
# Công thức Min-Max Normalization:
# US_TBM = (TBM / 10) * 4

# Tạo các biến US_TBM
df["US_TBM1"] = (df["TBM1"] / 10) * 4
df["US_TBM2"] = (df["TBM2"] / 10) * 4
df["US_TBM3"] = (df["TBM3"] / 10) * 4

# Làm tròn 2 chữ số thập phân
df["US_TBM1"] = df["US_TBM1"].round(2)
df["US_TBM2"] = df["US_TBM2"].round(2)
df["US_TBM3"] = df["US_TBM3"].round(2)

# Hiển thị kết quả 10 dòng đầu
print("KẾT QUẢ TẠO BIẾN US_TBM1, US_TBM2, US_TBM3")
display(df[[
    "TBM1", "US_TBM1",
    "TBM2", "US_TBM2",
    "TBM3", "US_TBM3"
]].head(10))

# Thống kê mô tả các biến mới
print("THỐNG KÊ MÔ TẢ CÁC BIẾN US_TBM")
display(df[["US_TBM1", "US_TBM2", "US_TBM3"]].describe())

print("Kết luận: Đã tạo thành công các biến US_TBM1, US_TBM2, US_TBM3 bằng phương pháp Min-Max Normalization.")
```

## Câu 10: Tạo biến KQXT - Kết quả xét tuyển

```python
# Quy tắc:
# Nếu khối A hoặc A1: điểm xét tuyển = (DH1*2 + DH2 + DH3) / 4
# Nếu khối B: điểm xét tuyển = (DH1 + DH2*2 + DH3) / 4
# Các khối còn lại: điểm xét tuyển = (DH1 + DH2 + DH3) / 3
# Nếu điểm xét tuyển >= 5 thì KQXT = 1, ngược lại KQXT = 0

def tinh_kqxt(row):
    khoi = str(row["KT"]).strip()
    dh1 = row["DH1"]
    dh2 = row["DH2"]
    dh3 = row["DH3"]
    
    if khoi in ["A", "A1"]:
        diem_xet_tuyen = (dh1 * 2 + dh2 + dh3) / 4
    elif khoi == "B":
        diem_xet_tuyen = (dh1 + dh2 * 2 + dh3) / 4
    else:
        diem_xet_tuyen = (dh1 + dh2 + dh3) / 3
    
    if diem_xet_tuyen >= 5:
        return 1
    else:
        return 0

# Tạo biến KQXT
df["KQXT"] = df.apply(tinh_kqxt, axis=1)

# Hiển thị kết quả 10 dòng đầu
print("KẾT QUẢ TẠO BIẾN KQXT")
display(df[["KT", "DH1", "DH2", "DH3", "KQXT"]].head(10))

# Thống kê tần số kết quả xét tuyển
print("Tần số kết quả xét tuyển:")
display(df["KQXT"].value_counts())

print("Kết luận: Đã tạo thành công biến KQXT.")
```

```python
# ==============================
# LAB 2 - PHẦN 3
# TRỰC QUAN HÓA DỮ LIỆU
# ==============================

# Import thư viện
import pandas as pd
import matplotlib.pyplot as plt

# Hiển thị đầy đủ bảng
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Upload file CSV
from google.colab import files
uploaded = files.upload()

# Đọc file
df = pd.read_csv('dulieuxettuyendaihoc (2).csv')

# ==========================================
# TẠO CÁC CỘT XL1, XL2, XL3
# ==========================================

def xeploai(diem):
    if diem < 5:
        return 'Y'
    elif diem < 6.5:
        return 'TB'
    elif diem < 8:
        return 'K'
    elif diem < 9:
        return 'G'
    else:
        return 'XS'

df['XL1'] = df['DH1'].apply(xeploai)
df['XL2'] = df['DH2'].apply(xeploai)
df['XL3'] = df['DH3'].apply(xeploai)

# ==========================================
# TẠO CỘT KQXT (Đậu / Rớt)
# ==========================================

def ketqua(row):
    if row['DH1'] >= 5 and row['DH2'] >= 5 and row['DH3'] >= 5:
        return 'Đậu'
    else:
        return 'Rớt'

df['KQXT'] = df.apply(ketqua, axis=1)

# ==========================================
# CÂU 1
# Học sinh nữ trên nhóm XL1 XL2 XL3
# ==========================================

print("===== CÂU 1 =====")

nu = df[df['GT'] == 'F']

bang_xl = pd.crosstab(
    index=['XL1', 'XL2', 'XL3'],
    columns='count'
)

# Tạo dữ liệu
xl1 = nu['XL1'].value_counts()
xl2 = nu['XL2'].value_counts()
xl3 = nu['XL3'].value_counts()

data = pd.DataFrame({
    'XL1': xl1,
    'XL2': xl2,
    'XL3': xl3
}).fillna(0)

data = data.reindex(['Y', 'TB', 'K', 'G', 'XS'])

# Vẽ biểu đồ
data.T.plot(kind='bar', stacked=False, figsize=(10,6))

plt.title('Học sinh nữ theo XL1 XL2 XL3')
plt.xlabel('Nhóm xếp loại')
plt.ylabel('Số lượng học sinh')

plt.show()


# ==========================================
# CÂU 2
# KQXT trên nhóm khối A A1 B thuộc KV 1,2
# ==========================================

print("===== CÂU 2 =====")

loc = df[
    (df['KT'].isin(['A', 'A1', 'B'])) &
    (df['KV'].isin(['1', '2']))
]

bang2 = pd.crosstab(loc['KT'], loc['KQXT'])

print(bang2)

bang2.plot(kind='bar', figsize=(8,5))

plt.title('KQXT theo khối thi')
plt.xlabel('Khối thi')
plt.ylabel('Số lượng')

plt.show()


# ==========================================
# CÂU 3
# Số lượng thí sinh từng KV theo khối thi
# ==========================================

print("===== CÂU 3 =====")

bang3 = pd.crosstab(df['KT'], df['KV'])

print(bang3)

bang3.plot(kind='bar', figsize=(10,6))

plt.title('Số lượng thí sinh theo KV và KT')
plt.xlabel('Khối thi')
plt.ylabel('Số lượng')

plt.show()


# ==========================================
# CÂU 4
# Đậu rớt theo nhóm khối thi
# ==========================================

print("===== CÂU 4 =====")

bang4 = pd.crosstab(df['KT'], df['KQXT'])

print(bang4)

bang4.plot(kind='bar', figsize=(8,5))

plt.title('Đậu rớt theo khối thi')
plt.xlabel('Khối thi')
plt.ylabel('Số lượng')

plt.show()


# ==========================================
# CÂU 5
# Đậu rớt theo khu vực
# ==========================================

print("===== CÂU 5 =====")

bang5 = pd.crosstab(df['KV'], df['KQXT'])

print(bang5)

bang5.plot(kind='bar', figsize=(8,5))

plt.title('Đậu rớt theo khu vực')
plt.xlabel('Khu vực')
plt.ylabel('Số lượng')

plt.show()


# ==========================================
# CÂU 6
# Đậu rớt theo dân tộc
# ==========================================

print("===== CÂU 6 =====")

# Điền NaN thành Kinh
df['DT'] = df['DT'].fillna('Kinh')

bang6 = pd.crosstab(df['DT'], df['KQXT'])

print(bang6)

bang6.plot(kind='bar', figsize=(10,6))

plt.title('Đậu rớt theo dân tộc')
plt.xlabel('Dân tộc')
plt.ylabel('Số lượng')

plt.show()


# ==========================================
# CÂU 7
# Đậu rớt theo giới tính
# ==========================================

print("===== CÂU 7 =====")

bang7 = pd.crosstab(df['GT'], df['KQXT'])

print(bang7)

bang7.plot(kind='bar', figsize=(6,5))

plt.title('Đậu rớt theo giới tính')
plt.xlabel('Giới tính')
plt.ylabel('Số lượng')

plt.show()
```

```python
# =====================================================================
# LAB 2 - PHẦN 4: TRỰC QUAN HÓA DỮ LIỆU NÂNG CAO (CHUẨN KẾT QUẢ SPSS)
# =====================================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập hiển thị dữ liệu đầy đủ trên Console
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# ---------------------------------------------------------------------
# ĐỌC DỮ LIỆU THỰC TẾ
# Bạn chỉ cần mở dòng code bên dưới ra nếu chạy trực tiếp bằng file CSV của bạn:
# df = pd.read_csv('dulieuxettuyendaihoc (2).csv')
# ---------------------------------------------------------------------

# Giả lập tập dữ liệu T1 để code chạy được độc lập (Bỏ qua nếu bạn đã load file CSV ở trên)
import numpy as np
np.random.seed(42)
t1_sample = np.round(np.random.uniform(2.4, 9.3, 100), 1)
df = pd.DataFrame({'T1': t1_sample})


# =====================================================================
# CÂU 1: BIỂU ĐỒ ĐƯỜNG SIMPLE CHO BIẾN T1 (TÍNH THEO TẦN SUẤT - COUNT)
# =====================================================================
print("\n===== CÂU 1 =====")

# Thống kê tần suất xuất hiện của từng mức điểm T1 và sắp xếp tăng dần
df_counts = df['T1'].value_counts().reset_index()
df_counts.columns = ['T1', 'Count']
df_counts = df_counts.sort_values(by='T1').reset_index(drop=True)

plt.figure(figsize=(11, 5))

# Vẽ đường Simple Line với các mức điểm phân bổ đều nhau trên trục hoành
plt.plot(df_counts['T1'].astype(str), df_counts['Count'], color='black', linewidth=1)

plt.xlabel('T1', fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.xticks(rotation=90, fontsize=8)
plt.ylim(0, df_counts['Count'].max() + 1)
plt.tight_layout()
plt.show()


# =====================================================================
# CÂU 2: TẠO BIẾN PHÂN LOẠI (phanlopt1) CHO MÔN TOÁN (T1)
# =====================================================================
print("\n===== CÂU 2 =====")

def phanloai(diem):
    if diem < 5:
        return 'k'    # Kém (Ký hiệu "k")
    elif diem < 7:
        return 'tb'   # Trung bình (Ký hiệu "tb")
    elif diem < 8:
        return 'kh'   # Khá (Ký hiệu "kh" để phân biệt rõ với "k" kém khi xử lý)
    else:
        return 'g'    # Giỏi (Ký hiệu "g")

# Tạo cột phanlopt1 theo đúng yêu cầu đề bài
df['phanlopt1'] = df['T1'].apply(phanloai)

# Hiển thị vài dòng đầu để kiểm tra kết quả
print(df[['T1', 'phanlopt1']].head(10))


# =====================================================================
# CÂU 3: LẬP BẢNG TẦN SỐ CHO BIẾN phanlopt1
# =====================================================================
print("\n===== CÂU 3 =====")

tan_so = df['phanlopt1'].value_counts()
print("Bảng tần số phân loại học lực T1:")
print(tan_so)


# =====================================================================
# CÂU 4: VẼ BIỂU ĐỒ ĐƯỜNG MULTIPLE LINE CHO T1 PHÂN LOẠI BỞI phanlopt1
# =====================================================================
print("\n===== CÂU 4 =====")

# Thống kê số lượng học sinh cho từng mức điểm thuộc từng nhóm học lực
df_multi = df.groupby(['T1', 'phanlopt1']).size().reset_index(name='Count')

# Thu thập toàn bộ các nhãn điểm duy nhất trên trục hoành phân loại để giữ khoảng cách đều
all_t1_str = sorted(df['T1'].unique().astype(str))

plt.figure(figsize=(11, 5))

# Bảng mã màu sắc và nhãn hiển thị mô phỏng theo mẫu đồ thị của SPSS
colors = {'g': '#4f81bd', 'kh': '#c0504d', 'k': '#9bbb59', 'tb': '#8064a2'}
labels = {'g': 'Gioi', 'kh': 'Kha', 'k': 'Kem', 'tb': 'Trungbinh'}

# Duyệt và vẽ từng nhóm độc lập để các đường nối liền mạch không bị ép kéo về đáy 0 lộn xộn
for group in ['g', 'k', 'kh', 'tb']:
    df_sub = df_multi[df_multi['phanlopt1'] == group].copy()

    if not df_sub.empty:
        df_sub = df_sub.sort_values(by='T1')
        plt.plot(df_sub['T1'].astype(str), df_sub['Count'],
                 label=labels[group], color=colors[group], linewidth=1.2)

plt.xlabel('T1', fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.xticks(ticks=all_t1_str, labels=all_t1_str, rotation=90, fontsize=8)
plt.ylim(0, df_multi['Count'].max() + 1)

# Định vị vị trí bảng chú thích (Legend) nằm ngoài đồ thị bên phải giống SPSS
plt.legend(title='Phanlopt1', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()


# =====================================================================
# CÂU 5: VẼ BIỂU ĐỒ DROP-LINE CHO T1 PHÂN LOẠI BỞI phanlopt1
# =====================================================================
print("\n===== CÂU 5 =====")

# Thống kê và sắp xếp dữ liệu gốc theo trật tự tăng dần của điểm số
df_drop = df.groupby(['T1', 'phanlopt1']).size().reset_index(name='Count')
df_drop = df_drop.sort_values(by='T1').reset_index(drop=True)
df_drop['T1_str'] = df_drop['T1'].astype(str)

plt.figure(figsize=(11, 5))

# 1. Khởi tạo các đường kẻ dọc (Drop-lines) hạ vuông góc từ vị trí điểm xuống trục đáy y=0
for idx, row in df_drop.iterrows():
    plt.vlines(x=row['T1_str'], ymin=0, ymax=row['Count'], colors='black', linewidth=1)

# 2. Đè các điểm Marker tròn có viền đen sắc nét lên đỉnh đầu của các đường kẻ dọc
sns.scatterplot(
    data=df_drop,
    x='T1_str',
    y='Count',
    hue='phanlopt1',
    hue_order=['g', 'k', 'kh', 'tb'],
    palette=colors,
    s=45,
    edgecolor='black',
    linewidth=0.8,
    zorder=3
)

# Định hình và đồng bộ hóa lại tên nhãn trong khung hiển thị chú thích nhóm
handles, labels_dict = plt.gca().get_legend_handles_labels()
new_labels = [labels.get(l, l) for l in labels_dict]

plt.xlabel('T1', fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.xticks(rotation=90, fontsize=8)
plt.ylim(0, df_drop['Count'].max() + 1)
plt.legend(handles, new_labels, title='Phanlopt1', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# =====================================================================
# LAB 2 - PHẦN 5: MÔ TẢ DỮ LIỆU VÀ KHẢO SÁT DẠNG PHÂN PHỐI
# =====================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Cấu hình hiển thị đầy đủ thông tin trên Console
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# ---------------------------------------------------------------------
# ĐỌC DỮ LIỆU THỰC TẾ (Bỏ chú thích dòng dưới nếu chạy file thực tế của bạn)
# df = pd.read_csv('dulieuxettuyendaihoc (2).csv')
# ---------------------------------------------------------------------

# Giả lập tập dữ liệu mẫu phục vụ chạy thử độc lập (Xóa đi nếu đã load file CSV)
np.random.seed(42)
n_samples = 100
t1_mock = np.round(np.random.normal(loc=5.5, scale=1.5, size=n_samples), 1)
t1_mock = np.clip(t1_mock, 0, 10)  # Giới hạn điểm từ 0 đến 10
dh1_mock = np.round(t1_mock * 0.6 + np.random.normal(loc=1.5, scale=0.8, size=n_samples), 1)
dh2_mock = np.round(t1_mock * 0.5 + np.random.normal(loc=2.0, scale=1.0, size=n_samples), 1)
dh3_mock = np.round(t1_mock * 0.55 + np.random.normal(loc=1.8, scale=0.9, size=n_samples), 1)

df = pd.DataFrame({'T1': t1_mock, 'DH1': dh1_mock, 'DH2': dh2_mock, 'DH3': dh3_mock})
def phanloai(diem):
    if diem < 5: return 'k'
    elif diem < 7: return 'tb'
    elif diem < 8: return 'kh'
    else: return 'g'
df['phanlopt1'] = df['T1'].apply(phanloai)


# =====================================================================
# CÂU 1: MÔ TẢ VÀ KHẢO SÁT PHÂN PHỐI CHO BIẾN T1
# =====================================================================
print("\n===== CÂU 1 =====")

# a. Mô tả độ tập trung, phân tán và các đặc trưng (lệch, nhọn)
mean_t1 = df['T1'].mean()
median_t1 = df['T1'].median()
mode_t1 = df['T1'].mode()[0]
var_t1 = df['T1'].var()
std_t1 = df['T1'].std()
skew_t1 = df['T1'].skew()
kurt_t1 = df['T1'].kurt()

print(f"--- Đặc trưng hình dáng & tập trung biến T1 ---")
print(f"Trung bình (Mean): {mean_t1:.4f}")
print(f"Trung vị (Median): {median_t1:.4f}")
print(f"Yếu vị (Mode): {mode_t1:.4f}")
print(f"Phương sai (Variance): {var_t1:.4f}")
print(f"Độ lệch chuẩn (Std Dev): {std_t1:.4f}")
print(f"Độ lệch (Skewness): {skew_t1:.4f} -> (Nếu gần 0: đối xứng, >0: lệch phải, <0: lệch trái)")
print(f"Độ nhọn (Kurtosis): {kurt_t1:.4f} -> (Nếu gần 0: phân phối chuẩn, >0: nhọn hơn, <0: dẹt hơn)")

# b. Vẽ biểu đồ Box-Plot và xác định 10 đại lượng
plt.figure(figsize=(6, 5))
sns.boxplot(y=df['T1'], color='#9bbb59')
plt.title('Biểu đồ Box-Plot của T1')
plt.ylabel('Điểm T1')
plt.show()

q1 = df['T1'].quantile(0.25)
q3 = df['T1'].quantile(0.75)
iqr = q3 - q1
min_val = df['T1'].min()
max_val = df['T1'].max()

print(f"\n--- 10 Đại lượng trong biểu đồ Box-Plot & dữ liệu ---")
print(f"1. Giá trị nhỏ nhất (Min): {min_val}")
print(f"2. Điểm phần vị thứ 25 (Q1): {q1}")
print(f"3. Điểm phần vị thứ 50 / Trung vị (Q2/Median): {median_t1}")
print(f"4. Điểm phần vị thứ 75 (Q3): {q3}")
print(f"5. Giá trị lớn nhất (Max): {max_val}")
print(f"6. Khoảng tứ phân vị (IQR): {iqr:.4f}")
print(f"7. Hàng rào dưới (Lower Fence): {q1 - 1.5*iqr:.4f}")
print(f"8. Hàng rào trên (Upper Fence): {q3 + 1.5*iqr:.4f}")
print(f"9. Số lượng mẫu (Count): {df['T1'].count()}")
print(f"10. Số lượng giá trị ngoại lai (Outliers): {df[(df['T1'] < (q1 - 1.5*iqr)) | (df['T1'] > (q3 + 1.5*iqr))]['T1'].count()}")

# c. Vẽ biểu đồ Histogram biểu thị hình dáng phân phối
plt.figure(figsize=(8, 5))
sns.histplot(df['T1'], kde=True, color='#4f81bd', bins=10)
plt.title('Biểu đồ Histogram với đường cong mật độ của T1')
plt.xlabel('Điểm T1')
plt.ylabel('Tần suất')
plt.show()

# d. Kiểm chứng phân phối chuẩn bằng QQ-Plot
plt.figure(figsize=(6, 5))
stats.probplot(df['T1'], plot=plt)
plt.title('Biểu đồ QQ-Plot kiểm chứng phân phối chuẩn của T1')
plt.show()


# =====================================================================
# CÂU 2: MÔ TẢ VÀ KHẢO SÁT PHÂN PHỐI T1 THEO TỪNG NHÓM PHÂN LỚP (phanlopt1)
# =====================================================================
print("\n===== CÂU 2 =====")

# Thống kê chi tiết T1 theo từng nhóm phân lớp
danh_gia_nhom = df.groupby('phanlopt1')['T1'].describe()
print("Thống kê mô tả T1 theo từng nhóm học lực:")
print(danh_gia_nhom)

# Trực quan hóa Box-plot theo phân nhóm
plt.figure(figsize=(8, 5))
sns.boxplot(x='phanlopt1', y='T1', data=df, palette='Set2', order=['k', 'tb', 'kh', 'g'])
plt.title('Biểu đồ Box-plot của T1 theo từng nhóm phân lớp')
plt.show()

# Trực quan hóa Histogram theo phân nhóm sử dụng FacetGrid
g = sns.FacetGrid(df, col="phanlopt1", col_order=['k', 'tb', 'kh', 'g'], height=4, aspect=1)
g.map(sns.histplot, "T1", kde=True, color='crimson')
g.set_titles("Nhóm: {col_name}")
plt.show()


# =====================================================================
# CÂU 3: KHẢO SÁT TƯƠNG QUAN GIỮA BIẾN DH1 THEO BIẾN T1
# =====================================================================
print("\n===== CÂU 3 =====")

# Nhận xét giá trị Covariance (Hiệp phương sai) và Correlation (Hệ số tương quan)
covariance_t1_dh1 = df['T1'].cov(df['DH1'])
correlation_t1_dh1 = df['T1'].corr(df['DH1'])

print(f"Hiệp phương sai (Covariance) giữa T1 và DH1: {covariance_t1_dh1:.4f}")
print(f"Hệ số tương quan Pearson (Correlation) giữa T1 và DH1: {correlation_t1_dh1:.4f}")
if correlation_t1_dh1 > 0:
    print("-> Nhận xét: T1 và DH1 có tương quan tuyến tính thuận (đồng biến).")
elif correlation_t1_dh1 < 0:
    print("-> Nhận xét: T1 và DH1 có tương quan tuyến tính nghịch (nghịch biến).")
else:
    print("-> Nhận xét: Không có tương quan tuyến tính giữa T1 và DH1.")

# Vẽ biểu đồ Scatter thể hiện mối liên hệ (DH1 là biến phụ thuộc -> Trục Y, T1 là biến độc lập -> Trục X)
plt.figure(figsize=(7, 5))
sns.scatterplot(x='T1', y='DH1', data=df, color='#c0504d', s=50, edgecolor='black')
# Thêm đường xu hướng tuyến tính (Trendline) để dễ quan sát mối quan hệ
sns.regplot(x='T1', y='DH1', data=df, scatter=False, color='black', label='Đường xu hướng')
plt.title('Biểu đồ Scatter biểu diễn mối quan hệ giữa DH1 theo T1')
plt.xlabel('Biến độc lập T1')
plt.ylabel('Biến phụ thuộc DH1')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


# =====================================================================
# CÂU 4: KHẢO SÁT TƯƠNG QUAN GIỮA DH1 THEO T1 TRÊN TỪNG NHÓM KHU VỰC
# =====================================================================
print("\n===== CÂU 4 =====")

# Lưu ý: Do trong tập mẫu hiện tại chưa có biến khu vực (KV), ta sẽ giả lập cột 'KV'
# để code chạy được trọn vẹn yêu cầu câu 4. (Nếu dùng file thật của bạn, hãy xóa dòng giả lập dưới đây).
if 'KV' not in df.columns:
    df['KV'] = np.random.choice(['KV1', 'KV2', 'KV3'], size=len(df))

# Tính toán hệ số tương quan theo từng nhóm khu vực
for name, group in df.groupby('KV'):
    corr_val = group['T1'].corr(group['DH1'])
    print(f"Hệ số tương quan giữa T1 và DH1 tại nhóm khu vực {name}: {corr_val:.4f}")

# Vẽ biểu đồ Scatter phân tách màu sắc hoặc phân ô theo từng khu vực
plt.figure(figsize=(8, 5))
sns.scatterplot(x='T1', y='DH1', hue='KV', data=df, palette='Dark2', s=60)
plt.title('Biểu đồ Scatter giữa DH1 theo T1 phân loại theo Khu vực')
plt.xlabel('Điểm T1')
plt.ylabel('Điểm DH1')
plt.grid(True)
plt.show()


# =====================================================================
# CÂU 5: KHẢO SÁT TƯƠNG QUAN GIỮA CÁC BIẾN DH1, DH2, DH3
# =====================================================================
print("\n===== CÂU 5 =====")

# Lập ma trận hiệp phương sai (Covariance Matrix)
ma_tran_hiep_phuong_sai = df[['DH1', 'DH2', 'DH3']].cov()
print("--- Ma trận hiệp phương sai (Covariance Matrix) ---")
print(ma_tran_hiep_phuong_sai)

# Lập ma trận tương quan (Correlation Matrix)
ma_tran_tuong_quan = df[['DH1', 'DH2', 'DH3']].corr()
print("\n--- Ma trận tương quan (Correlation Matrix) ---")
print(ma_tran_tuong_quan)

# Trực quan hóa ma trận tương quan bằng biểu đồ nhiệt Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(ma_tran_tuong_quan, annot=True, cmap='Blues', fmt=".4f", linewidths=0.5)
plt.title('Biểu đồ Heatmap ma trận tương quan giữa DH1, DH2, DH3')
plt.show()

# Vẽ biểu đồ Scatter giữa các cặp biến bằng PairPlot
print("\nĐang khởi tạo biểu đồ đa hướng Scatter giữa các cặp biến...")
sns.pairplot(df[['DH1', 'DH2', 'DH3']], diag_kind='kde', plot_kws={'color':'#8064a2'})
plt.suptitle('Biểu đồ Scatter Matrix giữa các biến DH1, DH2, DH3', y=1.02)
plt.show()
```

```python
# #Problem 3
# Định nghĩa hàm xử lý quy đổi cho từng dòng dữ liệu trong cột Weight
def convert_weight(x):
    if pd.isna(x):
        return x
    x = str(x).strip()
    if 'lbs' in x:
        # Loại bỏ chữ 'lbs' và chuyển về dạng số thực
        val = float(x.replace('lbs', '').strip())
        # Quy đổi sang kg và chuyển thành chuỗi kèm chữ 'kgs' như mẫu đề bài
        return f"{int(val / 2.2)}kgs"
    return x

# Áp dụng hàm quy đổi cho toàn bộ cột 'Weight'
df['Weight'] = df['Weight'].apply(convert_weight)

# Hiển thị kết quả sau khi đồng nhất đơn vị
print("--- Dữ liệu sau khi xử lý Vấn đề 3 ---")
display(df.head(10))
```

```python
# Đảm bảo cột PulseRate ở dạng số thực để tính toán toán học
df['PulseRate'] = pd.to_numeric(df['PulseRate'], errors='coerce')

# Sắp xếp lại index để thực hiện các phép toán liền trước / liền sau chính xác
df.reset_index(drop=True, inplace=True)

# Sao lưu dữ liệu để xử lý nội suy (liền trước/sau) bằng hàm ffill và bfill của pandas
df['forward_f'] = df.groupby(['Firstname', 'Lastname'])['PulseRate'].ffill()
df['backward_f'] = df.groupby(['Firstname', 'Lastname'])['PulseRate'].bfill()

# Tính toán các giá trị trung bình (Mean) dự phòng theo nhóm bài yêu cầu
mean_patient = df.groupby(['Firstname', 'Lastname'])['PulseRate'].transform('mean')
mean_sex = df.groupby('Sex')['PulseRate'].transform('mean')
mean_global = df['PulseRate'].mean()
medical_stable = 72.0 # Mức ổn định trong y học mặc định nếu tất cả đều trống

# Thực hiện điền dữ liệu thiếu theo thứ tự ưu tiên lồng nhau
def fill_pulse_rate(row):
    if not pd.isna(row['PulseRate']):
        return row['PulseRate']

    # 1. Trung bình liền trước và liền sau
    if not pd.isna(row['forward_f']) and not pd.isna(row['backward_f']):
        return (row['forward_f'] + row['backward_f']) / 2
    # 2. Giá trị liền trước
    if not pd.isna(row['forward_f']):
        return row['forward_f']
    # 3. Giá trị liền sau
    if not pd.isna(row['backward_f']):
        return row['backward_f']
    # 4. Trung bình của chính người đó
    if not pd.isna(mean_patient[row.name]):
        return mean_patient[row.name]
    # 5. Trung bình của nhóm giới tính
    if not pd.isna(mean_sex[row.name]):
        return mean_sex[row.name]
    # 6. Trung bình toàn bộ tập dữ liệu hoặc mức y học ổn định
    return mean_global if not pd.isna(mean_global) else medical_stable

# Áp dụng hàm điền dữ liệu
df['PulseRate'] = df.apply(fill_pulse_rate, axis=1)

# Loại bỏ các cột phụ trợ tạo ra trong quá trình tính toán
df.drop(['forward_f', 'backward_f'], axis=1, inplace=True)

# Xóa các dòng trống hoàn toàn nếu có phát sinh sau khi melt
df.dropna(subset=['PulseRate'], inplace=True)

# Reindex lại dữ liệu cho đẹp mắt
df.reset_index(drop=True, inplace=True)

# Hiển thị dữ liệu sạch hoàn chỉnh
print("--- Dữ liệu đã làm sạch hoàn toàn ---")
display(df)
```

# LAB 05: NHẬN DẠNG QUẦN ÁO GIÀY DÉP THỜI TRANG VỚI BỘ DỮ LIỆU FASHION-MNIST, DÙNG THƯ VIỆN PYTORCH HUẤN LUYỆN MÔ HÌNH TRÊN GPU CỦA FIT-LAB

```python
# =========================
# 1. IMPORT VÀ TẢI DỮ LIỆU
# =========================

from torchvision import datasets
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from torch.optim import SGD, Adam

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

data_folder = "~/data/FMNIST"

fmnist = datasets.FashionMNIST(
    data_folder,
    download=True,
    train=True
)

tr_images = fmnist.data
tr_targets = fmnist.targets

unique_values = tr_targets.unique()

print("tr_images & tr_targets:")
print("X -", tr_images.shape)
print("Y -", tr_targets.shape)
print("Y - Unique Values:", unique_values)
print("TASK:")
print(len(unique_values), "class Classification")
print("UNIQUE CLASSES:")
print(fmnist.classes)
```

```python
# =========================
# 2. HIỂN THỊ ẢNH MẪU
# =========================

R, C = len(tr_targets.unique()), 10

fig, ax = plt.subplots(R, C, figsize=(10, 10))

for label_class, plot_row in enumerate(ax):
    label_x_rows = np.where(tr_targets == label_class)[0]

    for plot_cell in plot_row:
        plot_cell.grid(False)
        plot_cell.axis("off")

        ix = np.random.choice(label_x_rows)
        x, y = tr_images[ix], tr_targets[ix]

        plot_cell.imshow(x, cmap="gray")

plt.tight_layout()
plt.show()
```

```python
# ======================================================
# 3. HUẤN LUYỆN MẠNG NƠ-RON - CHƯA CHIA ẢNH CHO 255
# ======================================================

class FMNISTDataset(Dataset):
    def __init__(self, x, y):
        x = x.float()
        x = x.view(-1, 28 * 28)
        self.x, self.y = x, y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)


def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)
    return trn_dl


def get_model():
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000),
        nn.ReLU(),
        nn.Linear(1000, 10)
    ).to(device)

    loss_fn = nn.CrossEntropyLoss()
    optimizer = SGD(model.parameters(), lr=1e-2)

    return model, loss_fn, optimizer


def train_batch(x, y, model, opt, loss_fn):
    model.train()

    prediction = model(x)
    batch_loss = loss_fn(prediction, y)

    batch_loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    return batch_loss.item()


@torch.no_grad()
def accuracy(x, y, model):
    model.eval()

    prediction = model(x)
    max_values, argmaxes = prediction.max(-1)

    is_correct = argmaxes == y

    return is_correct.cpu().numpy().tolist()


trn_dl = get_data()
model, loss_fn, optimizer = get_model()

losses, accuracies = [], []

for epoch in range(5):
    print("Epoch:", epoch + 1)

    epoch_losses, epoch_accuracies = [], []

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        batch_loss = train_batch(x, y, model, optimizer, loss_fn)
        epoch_losses.append(batch_loss)

    epoch_loss = np.array(epoch_losses).mean()

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        is_correct = accuracy(x, y, model)
        epoch_accuracies.extend(is_correct)

    epoch_accuracy = np.mean(epoch_accuracies)

    losses.append(epoch_loss)
    accuracies.append(epoch_accuracy)

    print("Loss:", epoch_loss)
    print("Accuracy:", epoch_accuracy)


epochs = np.arange(5) + 1

plt.figure(figsize=(20, 5))

plt.subplot(121)
plt.title("Training Loss")
plt.plot(epochs, losses, label="Training Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.subplot(122)
plt.title("Training Accuracy")
plt.plot(epochs, accuracies, label="Training Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.gca().set_yticklabels(["{:.0f}%".format(x * 100) for x in plt.gca().get_yticks()])
plt.legend()

plt.show()
```

```python
# ============================================
# TÁC ĐỘNG CỦA TỐC ĐỘ HỌC - FASHION MNIST
# ============================================

from torchvision import datasets
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
from torch.optim import Adam
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

# Tải dữ liệu
data_folder = "~/data/FMNIST"

fmnist = datasets.FashionMNIST(
    data_folder,
    download=True,
    train=True
)

tr_images = fmnist.data
tr_targets = fmnist.targets

val_fmnist = datasets.FashionMNIST(
    data_folder,
    download=True,
    train=False
)

val_images = val_fmnist.data
val_targets = val_fmnist.targets


# Dataset có chia tỷ lệ ảnh cho 255
class FMNISTDataset(Dataset):
    def __init__(self, x, y):
        x = x.float() / 255
        x = x.view(-1, 28 * 28)
        self.x = x
        self.y = y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)


def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)

    val = FMNISTDataset(val_images, val_targets)
    val_dl = DataLoader(val, batch_size=len(val_images), shuffle=False)

    return trn_dl, val_dl


def get_model(lr):
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000),
        nn.ReLU(),
        nn.Linear(1000, 10)
    ).to(device)

    loss_fn = nn.CrossEntropyLoss()
    optimizer = Adam(model.parameters(), lr=lr)

    return model, loss_fn, optimizer


def train_batch(x, y, model, opt, loss_fn):
    model.train()

    prediction = model(x)
    batch_loss = loss_fn(prediction, y)

    batch_loss.backward()
    opt.step()
    opt.zero_grad()

    return batch_loss.item()


@torch.no_grad()
def accuracy(x, y, model):
    model.eval()

    prediction = model(x)
    max_values, argmaxes = prediction.max(-1)

    is_correct = argmaxes == y

    return is_correct.cpu().numpy().tolist()


@torch.no_grad()
def val_loss(x, y, model, loss_fn):
    model.eval()

    prediction = model(x)
    validation_loss = loss_fn(prediction, y)

    return validation_loss.item()


def train_model(lr, epochs_count=5):
    trn_dl, val_dl = get_data()
    model, loss_fn, optimizer = get_model(lr)

    train_losses, train_accuracies = [], []
    val_losses, val_accuracies = [], []

    for epoch in range(epochs_count):
        print("Epoch:", epoch + 1)

        train_epoch_losses = []
        train_epoch_accuracies = []

        # Huấn luyện
        for ix, batch in enumerate(iter(trn_dl)):
            x, y = batch
            batch_loss = train_batch(x, y, model, optimizer, loss_fn)
            train_epoch_losses.append(batch_loss)

        train_epoch_loss = np.array(train_epoch_losses).mean()

        # Accuracy trên tập train
        for ix, batch in enumerate(iter(trn_dl)):
            x, y = batch
            is_correct = accuracy(x, y, model)
            train_epoch_accuracies.extend(is_correct)

        train_epoch_accuracy = np.mean(train_epoch_accuracies)

        # Validation
        for ix, batch in enumerate(iter(val_dl)):
            x, y = batch
            val_is_correct = accuracy(x, y, model)
            validation_loss = val_loss(x, y, model, loss_fn)

        val_epoch_accuracy = np.mean(val_is_correct)

        train_losses.append(train_epoch_loss)
        train_accuracies.append(train_epoch_accuracy)
        val_losses.append(validation_loss)
        val_accuracies.append(val_epoch_accuracy)

        print("Train Loss:", train_epoch_loss)
        print("Train Accuracy:", train_epoch_accuracy)
        print("Validation Loss:", validation_loss)
        print("Validation Accuracy:", val_epoch_accuracy)
        print("-" * 40)

    return train_losses, train_accuracies, val_losses, val_accuracies


def plot_result(train_losses, train_accuracies, val_losses, val_accuracies, lr):
    epochs = np.arange(len(train_losses)) + 1

    plt.figure(figsize=(10, 6))

    plt.subplot(211)
    plt.plot(epochs, train_losses, "bo", label="Training loss")
    plt.plot(epochs, val_losses, "r", label="Validation loss")
    plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
    plt.title(f"Training and validation loss with {lr} learning rate")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(False)

    plt.subplot(212)
    plt.plot(epochs, train_accuracies, "bo", label="Training accuracy")
    plt.plot(epochs, val_accuracies, "r", label="Validation accuracy")
    plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
    plt.title(f"Training and validation accuracy with {lr} learning rate")
    plt.xlabel("Epochs")
    plt.ylabel("Accuracy")
    plt.gca().set_yticklabels(["{:.0f}%".format(x * 100) for x in plt.gca().get_yticks()])
    plt.legend()
    plt.grid(False)

    plt.tight_layout()
    plt.show()
```

```python
train_losses, train_accuracies, val_losses, val_accuracies = train_model(
    lr=1e-1,
    epochs_count=5
)

plot_result(
    train_losses,
    train_accuracies,
    val_losses,
    val_accuracies,
    lr=0.1
)
```

```python
# ============================================================
# TÁC ĐỘNG CỦA TỐC ĐỘ HỌC TRÊN DỮ LIỆU KHÔNG CHIA /255
# ============================================================

from torchvision import datasets
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
from torch.optim import Adam
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

data_folder = "./data/FMNIST"

fmnist = datasets.FashionMNIST(data_folder, download=True, train=True)
tr_images = fmnist.data
tr_targets = fmnist.targets

val_fmnist = datasets.FashionMNIST(data_folder, download=True, train=False)
val_images = val_fmnist.data
val_targets = val_fmnist.targets


# Dataset KHÔNG chia cho 255
class FMNISTDataset(Dataset):
    def __init__(self, x, y):
        x = x.float()
        x = x.view(-1, 28 * 28)
        self.x = x
        self.y = y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)


def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)

    val = FMNISTDataset(val_images, val_targets)
    val_dl = DataLoader(val, batch_size=len(val_images), shuffle=False)

    return trn_dl, val_dl


def get_model(lr):
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000),
        nn.ReLU(),
        nn.Linear(1000, 10)
    ).to(device)

    loss_fn = nn.CrossEntropyLoss()
    optimizer = Adam(model.parameters(), lr=lr)

    return model, loss_fn, optimizer


def train_batch(x, y, model, optimizer, loss_fn):
    model.train()

    prediction = model(x)
    batch_loss = loss_fn(prediction, y)

    batch_loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    return batch_loss.item()


@torch.no_grad()
def accuracy(x, y, model):
    model.eval()

    prediction = model(x)
    max_values, argmaxes = prediction.max(-1)
    is_correct = argmaxes == y

    return is_correct.cpu().numpy().tolist()


@torch.no_grad()
def val_loss(x, y, model, loss_fn):
    model.eval()

    prediction = model(x)
    validation_loss = loss_fn(prediction, y)

    return validation_loss.item()


def run_lr_experiment(lr, epochs_count=5):
    trn_dl, val_dl = get_data()
    model, loss_fn, optimizer = get_model(lr)

    train_losses, train_accuracies = [], []
    val_losses, val_accuracies = [], []

    for epoch in range(epochs_count):
        print("Learning rate:", lr, "- Epoch:", epoch)

        train_epoch_losses, train_epoch_accuracies = [], []

        for ix, batch in enumerate(iter(trn_dl)):
            x, y = batch
            batch_loss = train_batch(x, y, model, optimizer, loss_fn)
            train_epoch_losses.append(batch_loss)

        train_epoch_loss = np.array(train_epoch_losses).mean()

        for ix, batch in enumerate(iter(trn_dl)):
            x, y = batch
            is_correct = accuracy(x, y, model)
            train_epoch_accuracies.extend(is_correct)

        train_epoch_accuracy = np.mean(train_epoch_accuracies)

        for ix, batch in enumerate(iter(val_dl)):
            x, y = batch
            val_is_correct = accuracy(x, y, model)
            validation_loss = val_loss(x, y, model, loss_fn)

        val_epoch_accuracy = np.mean(val_is_correct)

        train_losses.append(train_epoch_loss)
        train_accuracies.append(train_epoch_accuracy)
        val_losses.append(validation_loss)
        val_accuracies.append(val_epoch_accuracy)

    return model, train_losses, train_accuracies, val_losses, val_accuracies
```

```python
# =====================================================================
# BÀI TẬP THỰC HÀNH MÔN NHẬP MÔN PHÂN TÍCH DỮ LIỆU VÀ HỌC SÂU
# LAB 6: XÂY DỰNG MẠNG NƠ-RON SÂU TOÀN DIỆN (FASHION-MNIST)
# Môi trường chạy: Chọn Runtime -> Change runtime type -> T4 GPU
# =====================================================================

import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

print("=====================================================================")
print("1. KIỂM TRA MÔI TRƯỜNG HỆ THỐNG")
print("=====================================================================")
print("Phiên bản TensorFlow hiện tại:", tf.__version__)
print("Danh sách thiết bị phần cứng tăng tốc (GPU):")
gpu_devices = tf.config.list_physical_devices('GPU')
print(gpu_devices if gpu_devices else "Không tìm thấy GPU - Vui lòng bật T4 GPU trong cài đặt Runtime!")

# ---------------------------------------------------------------------
# Bước 1: Tải và Tiền xử lý dữ liệu (Scaling dữ liệu)
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("2. TIẾN HÀNH TẢI VÀ CHUẨN HÓA DỮ LIỆU FASHION-MNIST")
print("=====================================================================")
fashion_mnist = keras.datasets.fashion_mnist
(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

print(f"Kích thước tập huấn luyện gốc (Train Full): {X_train_full.shape}")
print(f"Kích thước tập kiểm tra gốc (Test): {X_test.shape}")

# Tách tập Validation (5000 ảnh đầu) và chuẩn hóa pixel về khoảng [0, 1]
X_valid, X_train = X_train_full[:5000] / 255.0, X_train_full[5000:] / 255.0
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]
X_test = X_test / 255.0

# Định nghĩa tên các nhãn tương ứng với các ID từ 0 đến 9
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

# ---------------------------------------------------------------------
# Bước 2: Thiết kế cấu trúc Mạng Nơ-ron Sâu (DNN)
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("3. THIẾT KẾ CẤU TRÚC MẠNG NƠ-RON SÂU (DNN MULTI-LAYER PERCEPTRON)")
print("=====================================================================")
model = keras.models.Sequential([
    # Lớp Flatten làm phẳng ma trận ảnh 2D (28x28) thành mảng 1D (784 nơ-ron)
    keras.layers.Flatten(input_shape=[28, 28]),

    # Lớp ẩn thứ 1 (Dense): 300 nơ-ron kết hợp hàm kích hoạt ReLU
    keras.layers.Dense(300, activation="relu"),

    # Lớp ẩn thứ 2 (Dense): 100 nơ-ron kết hợp hàm kích hoạt ReLU
    keras.layers.Dense(100, activation="relu"),

    # Lớp đầu ra (Dense): 10 nơ-ron đại diện cho 10 lớp, dùng Softmax tính xác suất
    keras.layers.Dense(10, activation="softmax")
])

# Hiển thị thông số cấu trúc phân lớp mạng sâu
model.summary()

# ---------------------------------------------------------------------
# Bước 3: Biên dịch và Huấn luyện mô hình
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("4. BIÊN DỊCH VÀ TIẾN HÀNH HUẤN LUYỆN MÔ HÌNH (TRAINING)")
print("=====================================================================")
model.compile(loss="sparse_categorical_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])

# Tiến hành traning trong 20 vòng Epochs
history = model.fit(X_train, y_train, epochs=20,
                    validation_data=(X_valid, y_valid))

# ---------------------------------------------------------------------
# Bước 4: Đánh giá hiệu năng và xuất báo cáo phân loại nâng cao
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("5. ĐÁNH GIÁ ĐỘ CHÍNH XÁC TOÀN DIỆN TRÊN TẬP KIỂM TRA (TEST SET)")
print("=====================================================================")
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"=> Kết quả thu được - Test Loss: {test_loss:.4f}")
print(f"=> Kết quả thu được - Test Accuracy (Độ chính xác tổng thể): {test_acc*100:.2f}%\n")

# Dự đoán xác suất cho toàn bộ tập kiểm tra
y_proba = model.predict(X_test, verbose=0)
y_pred = np.argmax(y_proba, axis=1)

print("\n--- BÁO CÁO PHÂN LOẠI CHI TIẾT (CLASSIFICATION REPORT) ---")
print(classification_report(y_test, y_pred, target_names=class_names))

# ---------------------------------------------------------------------
# Bước 5: Hiển thị trực quan Ma trận nhầm lẫn (Confusion Matrix Heatmap)
# ---------------------------------------------------------------------
print("\n=====================================================================")
# Trực quan hóa ma trận biểu diễn các lỗi nhầm lẫn của mô hình nhận dạng ảnh
print("6. TRỰC QUAN HÓA MA TRẬN NHẦM LẪN (CONFUSION MATRIX)")
print("=====================================================================")
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - Lab 6 Fashion-MNIST', fontsize=12, fontweight='bold')
plt.ylabel('Nhãn thực tế (True Labels)')
plt.xlabel('Nhãn dự đoán (Predicted Labels)')
plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------
# Bước 6: Trực quan hóa đồ thị Lịch sử học tập (Loss & Accuracy Plots)
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("7. ĐỒ THỊ BIỂU DIỄN TIẾN TRÌNH HỌC TẬP QUA 20 EPOCHS")
print("=====================================================================")
epochs_range = range(1, len(history.history['accuracy']) + 1)
plt.figure(figsize=(14, 5))

# Đồ thị biểu diễn giá trị Loss giảm dần
plt.subplot(1, 2, 1)
plt.plot(epochs_range, history.history['loss'], 'bo-', label='Training Loss')
plt.plot(epochs_range, history.history['val_loss'], 'r-', label='Validation Loss')
plt.title('Loss Value over Increasing Epochs', fontsize=11, fontweight='bold')
plt.xlabel('Epochs')
plt.ylabel('Loss Value')
plt.legend()
plt.grid(True)

# Đồ thị biểu diễn giá trị Accuracy tăng dần
plt.subplot(1, 2, 2)
plt.plot(epochs_range, history.history['accuracy'], 'bo-', label='Training Accuracy')
plt.plot(epochs_range, history.history['val_accuracy'], 'r-', label='Validation Accuracy')
plt.title('Accuracy Value over Increasing Epochs', fontsize=11, fontweight='bold')
plt.xlabel('Epochs')
plt.ylabel('Accuracy (%)')
plt.gca().set_yticklabels(['{:.0f}%'.format(x*100) for x in plt.gca().get_yticks()])
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------
# Bước 7: Hiển thị HOÀNH TRÁNG lưới 20 kết quả dự đoán thực tế (4 hàng x 5 cột)
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("8. TRỰC QUAN HÓA LƯỚI ĐẦY ĐỦ 20 KẾT QUẢ CHẠY THỰC TẾ TRÊN MÀN HÌNH")
print("=====================================================================")
plt.figure(figsize=(15, 12))

for i in range(20):
    plt.subplot(4, 5, i + 1)
    plt.imshow(X_test[i], cmap="binary")
    plt.axis('off')

    # Text màu XANH LÁ nếu dự đoán đúng, màu ĐỎ nếu dự đoán bị lệch
    color = "green" if y_pred[i] == y_test[i] else "red"
    confidence = round(float(np.max(y_proba[i])) * 100, 2)

    plt.title(f"Pred: {class_names[y_pred[i]]}\nTrue: {class_names[y_test[i]]}\nConf: {confidence}%",
              color=color, fontsize=10)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------------------
# Bước 8: Xổ bảng dữ liệu thực nghiệm FULL 100 dòng ngay trên màn hình Colab
# ---------------------------------------------------------------------
print("\n=====================================================================")
print("9. BẢNG THỐNG KÊ CHI TIẾT ĐỦ ĐÚNG 100 DÒNG KẾT QUẢ DỰ ĐOÁN")
print("=====================================================================")
so_mau = 100
danh_sach_mau = range(so_mau)

data_100_dong = {
    'STT': [i + 1 for i in danh_sach_mau],
    'Mẫu ảnh số (Index)': danh_sach_mau,
    'Nhãn thực tế (True ID)': y_test[danh_sach_mau],
    'Tên nhãn gốc (True Name)': [class_names[i] for i in y_test[danh_sach_mau]],
    'Nhãn dự đoán (Pred ID)': y_pred[danh_sach_mau],
    'Tên nhãn đoán (Pred Name)': [class_names[i] for i in y_pred[danh_sach_mau]],
    'Độ tự tin mạng (%)': [round(float(np.max(y_proba[i])) * 100, 2) for i in danh_sach_mau],
    'Đánh giá': ['Đúng (Chính xác)' if y_pred[i] == y_test[i] else 'Sai (Nhầm lẫn)' for i in danh_sach_mau]
}

df_lab06_100_dong = pd.DataFrame(data_100_dong)

# Cấu hình hiển thị đầy đủ không bị ẩn dòng hay giấu bớt thông tin
pd.set_option('display.max_rows', 105)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# In toàn bộ 100 dòng dữ liệu ra màn hình Colab
print(df_lab06_100_dong.to_string(index=False))
print("\n=====================================================================")
print("HOÀN THÀNH TOÀN BỘ TIẾN TRÌNH THỰC HÀNH LAB 6!")
print("=====================================================================")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download dữ liệu
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

text = "This is a simple example to show how to remove stopwords from a sentence."

stop_words = set(stopwords.words("english"))

words = word_tokenize(text)

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("Văn bản ban đầu:")
print(text)

print("\nSau khi loại bỏ stopwords:")
print(filtered_words)
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

word = "good"

synonyms = []
antonyms = []

for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())

        if lemma.antonyms():
            antonyms.append(lemma.antonyms()[0].name())

print("Từ cần tìm:", word)

print("\nTừ đồng nghĩa:")
print(set(synonyms))

print("\nTừ trái nghĩa:")
print(set(antonyms))
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

word = "dog"

synsets = wordnet.synsets(word)

syn = synsets[0]

print("Từ:", word)
print("Synset:", syn.name())
print("Định nghĩa:", syn.definition())

print("\nHypernyms - từ khái quát hơn:")
for h in syn.hypernyms():
    print(h.name(), ":", h.definition())

print("\nHyponyms - từ cụ thể hơn:")
for h in syn.hyponyms()[:10]:
    print(h.name(), ":", h.definition())
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

word = "happy"

synsets = wordnet.synsets(word)

print("Từ:", word)
print("Số lượng synset:", len(synsets))

for i, syn in enumerate(synsets, start=1):
    print("\nSynset", i)
    print("Tên:", syn.name())
    print("Loại từ:", syn.pos())
    print("Định nghĩa:", syn.definition())
    print("Ví dụ:", syn.examples())
    print("Lemma:", syn.lemma_names())
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

verb1 = "run"
verb2 = "walk"

syn1 = wordnet.synsets(verb1, pos=wordnet.VERB)[0]
syn2 = wordnet.synsets(verb2, pos=wordnet.VERB)[0]

similarity = syn1.path_similarity(syn2)

print("Động từ 1:", verb1, "-", syn1.name())
print("Động từ 2:", verb2, "-", syn2.name())
print("Độ tương đồng:", similarity)
```

```python
import nltk
from nltk.corpus import names

# Tải dữ liệu
nltk.download('names')

# Đọc danh sách tên
male_names = names.words('male.txt')
female_names = names.words('female.txt')

# Hiển thị số lượng
print("Số lượng tên nam:", len(male_names))
print("Số lượng tên nữ:", len(female_names))

# Hiển thị 10 tên đầu tiên
print("\n10 tên nam đầu tiên:")
print(male_names[:10])

print("\n10 tên nữ đầu tiên:")
print(female_names[:10])

# Tìm tên xuất hiện ở cả nam và nữ
common_names = set(male_names).intersection(set(female_names))

print("\nSố tên xuất hiện ở cả nam và nữ:", len(common_names))

print("\n20 tên trùng nhau đầu tiên:")
print(sorted(common_names)[:20])
```
